# Tiền xử lý dữ liệu
Bộ dữ liệu: **Bike Sharing Dataset** — file `hour.csv`

```python
from pathlib import Path
from sklearn.model_selection import train_test_split
import pandas as pd

# Xác định ROOT: thư mục chứa 'data/'
try:
    ROOT = Path(__file__).resolve().parent.parent
except NameError:
    cwd = Path.cwd()
    for candidate in [cwd, cwd.parent, cwd.parent.parent]:
        if (candidate / "data").exists():
            ROOT = candidate
            break
    else:
        raise FileNotFoundError(f"Không tìm thấy thư mục 'data/' từ {cwd}")

DATA_DIR = ROOT / "data" / "regression"
IMGS_DIR = Path(Path.cwd()) / "imgs"
IMGS_DIR.mkdir(exist_ok=True)

print(f"ROOT     : {ROOT}")
print(f"DATA_DIR : {DATA_DIR}")
print(f"IMGS_DIR : {IMGS_DIR}")
assert (DATA_DIR / "hour.csv").exists(), f"Không tìm thấy {DATA_DIR / 'hour.csv'}"
```

## Missing values

Bộ dữ liệu không có giá trị thiếu, do đó không cần thực hiện bước xử lý missing values.

## Chuẩn hóa

Bộ dữ liệu đã được normalize, theo [mô tả của tác giả](../../data/Readme.txt).

Các biến `temp`, `atemp`, `hum`, `windspeed` đã được chuẩn hóa về khoảng [0, 1]. Do đó, không cần thực hiện bước chuẩn hóa thêm nào nữa.

## Chia dữ liệu

Ta chia dữ liệu theo tỉ lệ 70% train, 10% validation, 20% test.

**Quan trọng**: Vì dữ liệu là time series, ta không shuffle mà chia theo thứ tự thời gian để tránh data leakage.

```python
# Split
df = pd.read_csv(DATA_DIR / "hour.csv")
train_val_df, test_df = train_test_split(df, test_size=0.2, random_state=42, shuffle=False)
train_df, val_df = train_test_split(train_val_df, test_size=0.125, random_state=42, shuffle=False)  # 0.125 * 0.8 = 0.1
```

```python
# Save

# train
train_df.to_csv(DATA_DIR / "train.csv", index=False)

# val
val_df.to_csv(DATA_DIR / "val.csv", index=False)

# test
test_df.to_csv(DATA_DIR / "test.csv", index=False)
```
